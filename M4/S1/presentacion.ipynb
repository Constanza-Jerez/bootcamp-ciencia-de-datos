```python

### **Estructura de Ejemplo Python para Presentación de Fundamentos de Ciencia de Datos: Análisis Exploratorio de Datos (EDA)**

# =============================================================================
# 1: Introducción al Análisis Exploratorio de Datos (EDA)
# =============================================================================

# El Análisis Exploratorio de Datos (EDA) es una fase fundamental en cualquier
# proyecto de ciencia de datos.
# Consiste en la exploración inicial de los datos para comprender su estructura,
# identificar patrones, detectar anomalías y formular hipótesis.
# Este proceso es crucial para garantizar la calidad de los análisis posteriores
# y la validez de los modelos predictivos.


# =============================================================================
# 2: Objetivos del EDA
# =============================================================================

# Los principales objetivos del EDA incluyen:
# 1.  Comprender la estructura y distribución de los datos.
# 2.  Detectar valores atípicos (outliers) y datos faltantes.
# 3.  Descubrir relaciones entre variables.
# 4.  Formular hipótesis para orientar análisis futuros.

# =============================================================================
# 3: Análisis Inicial de Datos (IDA)
# =============================================================================

# El Análisis Inicial de Datos (IDA) es una fase preliminar del EDA.
# Se enfoca en la inspección básica de los datos para asegurar que estén listos
# para análisis más profundos.

print("--- Módulo 3: Análisis Inicial de Datos (IDA) ---")
print("El IDA es la inspección básica de los datos para su preparación.")

# Importar las bibliotecas necesarias (Pandas es clave para IDA)
import pandas as pd
import numpy as np # También útil, aunque no directamente en el ejemplo IDA

# --- Paso 1: Carga de datos ---
# Importar el conjunto de datos desde una fuente (ej. CSV).
# Es crucial verificar que los datos se carguen correctamente.
try:
    df = pd.read_csv('datos_ejemplo.csv') # Reemplaza con la ruta a tu archivo
    print("\n--- Paso 1: Carga de datos ---")
    print("Datos cargados exitosamente desde 'datos_ejemplo.csv'.")
except FileNotFoundError:
    print("\n--- Paso 1: Carga de datos (Simulada) ---")
    print("Archivo 'datos_ejemplo.csv' no encontrado. Creando un DataFrame de ejemplo.")
    # DataFrame de ejemplo
    data = {
        'ID': range(1, 11),
        'Edad': [23, 35, 29, 41, 27, 32, 38, 30, 26, 40],
        'Ingresos': [25000, 40000, 32000, 50000, 27000, 35000, 42000, 48000, 30000, 52000],
        'Educacion': ['Básica', 'Licenciatura', 'Maestría', 'Doctorado', 'Básica', 'Licenciatura', 'Maestría', 'Doctorado', 'Licenciatura', 'Maestría'],
        'Ciudad': ['A', 'B', 'A', 'C', 'B', 'A', 'C', 'B', 'A', 'C'],
        'Horas_Estudio': [5, 8, 10, 12, 6, 7, 9, 11, np.nan, 15], # np.nan para valores faltantes
        'Calificacion': [70, 85, 90, 95, 75, 80, 88, 92, 78, 98]
    }
    df = pd.DataFrame(data)
    # Añadir un duplicado para demostrar df.duplicated()
    df.loc[len(df)] = df.iloc[0]

# --- Paso 2: Inspección inicial ---
# Revisar las primeras filas, tipos de datos y dimensiones.
print("\n--- Paso 2: Inspección inicial ---")
print("Primeras 5 filas del dataset (df.head()):")
print(df.head()) # Muestra las primeras 5 filas
print("\nInformación del dataset (df.info()):")
df.info() # Muestra tipos de datos y dimensiones

# --- Paso 3: Identificación de problemas ---
# Detectar valores faltantes, duplicados o inconsistencias.

print("\n--- Paso 3: Identificación de problemas ---")
print("a) Valores faltantes por columna (df.isnull().sum()):")
print(df.isnull().sum()) # Muestra el número de valores faltantes por columna

print("\nb) Filas duplicadas en el dataset (df.duplicated().sum()):")
print(df.duplicated().sum()) # Muestra la cantidad de filas duplicadas

print("\nc) Estadísticas descriptivas para detectar valores atípicos (df.describe()):")
# df.describe() proporciona una vista general de estadísticas que pueden ayudar a detectar valores atípicos.
# Esta función incluye los cuartiles (25%, 50% que es la mediana, y 75% percentiles)
# junto con la media, desviación estándar, valores mínimo y máximo, y el conteo de datos.
# Aunque las fuentes no definen explícitamente "cuartiles", df.describe() es la herramienta
# mencionada que los calcula como parte de las "estadísticas descriptivas".
print(df.describe()) # Muestra estadísticas como media, desviación estándar, etc.

print("\nd) Tipos de datos para revisar posibles conversiones (df.dtypes):")
print(df.dtypes) # Muestra los tipos de datos de cada columna
print("\n")

# =============================================================================
# 4: Técnicas Clave del EDA
# =============================================================================

# Importar bibliotecas para visualización
from matplotlib import pyplot as plt
import seaborn as sns

print("--- Módulo 4: Técnicas Clave del EDA ---")

# --- 4.1 Análisis Univariado ---
# Propósito: Estudia una sola variable a la vez para entender su distribución y comportamiento.
# Objetivos: Comprender la distribución, detectar valores atípicos, evaluar centralidad y analizar dispersión.
print("\n--- 4.1 Análisis Univariado ---")
print("Estudia una sola variable para comprender su distribución y comportamiento.")

# Ejemplo con Pandas:
# Calcular medidas estadísticas
print("\nMedidas de Tendencia Central y Dispersión para 'Ingresos':")
print(f"Media de Ingresos: {df['Ingresos'].mean()}") # Calcula la media
print(f"Mediana de Ingresos: {df['Ingresos'].median()}") # Calcula la mediana (que es el Segundo Cuartil o Q2)
print(f"Desviación Estándar de Ingresos: {df['Ingresos'].std()}") # Calcula la desviación estándar

# Visualización con boxplot
# Los gráficos de caja (boxplots) son una herramienta clave en el análisis univariado.
# Visualizan los cuartiles, la mediana y la presencia de valores atípicos.
# La "caja" representa el rango intercuartílico (IQR), los "bigotes" se extienden
# hasta el valor más lejano que no es un valor atípico, y los puntos fuera de los bigotes
# son considerados atípicos. Esto nos ayuda a "detectar valores atípicos" y a "analizar dispersión".
plt.figure(figsize=(8, 4))
sns.boxplot(x=df["Ingresos"]) # Visualización con boxplot
plt.title("Boxplot de Ingresos") # Título del gráfico
plt.xlabel("Ingresos")
plt.show() # Muestra el gráfico
print("Boxplot generado para 'Ingresos' para visualizar distribución, centralidad, dispersión y atípicos.")

# Visualización con Histograma (mencionado en fuentes pero no con código)
plt.figure(figsize=(8, 4))
sns.histplot(df["Ingresos"], kde=True)
plt.title("Histograma de Ingresos")
plt.xlabel("Ingresos")
plt.ylabel("Frecuencia")
plt.show()
print("Histograma generado para 'Ingresos' para analizar su distribución.")
print("\n")


# --- 4.2 Análisis Bivariado ---
# Propósito: Analiza dos variables juntas para detectar relaciones o patrones.
# Herramientas: Diagramas de dispersión (scatter plots), correlación de Pearson y tablas cruzadas (crosstabs).
print("--- 4.2 Análisis Bivariado ---")
print("Analiza dos variables para detectar relaciones o patrones.")
print("Herramientas: Diagramas de dispersión, correlación de Pearson y tablas cruzadas.")

# Ejemplo con Pandas: Diagrama de Dispersión
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df['Edad'], y=df['Ingresos'])
plt.title("Diagrama de Dispersión: Edad vs Ingresos")
plt.xlabel("Edad")
plt.ylabel("Ingresos")
plt.show()
print("Diagrama de dispersión generado para 'Edad' vs 'Ingresos' para visualizar su relación.")

# Ejemplo con Pandas: Correlación de Pearson
# Aunque la correlación de Pearson se menciona en bivariado, su cálculo se
# ejemplifica en multivariado con la matriz de correlación.
print(f"\nCorrelación de Pearson entre Edad e Ingresos: {df['Edad'].corr(df['Ingresos'])}")
print("Correlación de Pearson calculada para 'Edad' e 'Ingresos'.")
print("\n")


# --- 4.3 Análisis Multivariado ---
# Propósito: Estudia la relación entre múltiples variables simultáneamente.
# Es útil para evaluar cómo una variable depende de otra y detectar tendencias en conjuntos de datos complejos.
print("--- 4.3 Análisis Multivariado ---")
print("Examina múltiples variables simultáneamente para identificar patrones y relaciones.")

# Técnicas de Análisis Multivariado:
# - Correlación: Evalúa la relación entre dos variables numéricas, usando el coeficiente de correlación de Pearson.
# - Análisis de Componentes Principales (PCA): Técnica de reducción de dimensionalidad.
# - Gráficos de Dispersión Matricial (Pair Plots): Visualización de relaciones entre pares de variables en una sola figura.

# Ejemplo de Análisis Multivariado:

# 1. Matriz de Correlación
# Para identificar las relaciones lineales entre variables.
print("\nMatriz de Correlación (df.corr()):")
print(df.corr(numeric_only=True)) # numeric_only=True para evitar errores con columnas no numéricas
print("La matriz de correlación proporciona una vista numérica clara de las relaciones lineales.")

# 2. Visualización con Mapa de Calor
# Se utiliza para representar gráficamente las correlaciones, facilitando la interpretación visual.
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm", fmt=".2f") # Representación gráfica de correlaciones
plt.title("Matriz de Correlación Visualizada con Mapa de Calor")
plt.show()
print("Mapa de calor generado para la matriz de correlación, útil para visualizar la fuerza de las relaciones.")

# 3. Análisis de Pares de Variables con Pair Plot
# Se emplea para explorar visualmente las relaciones bivariadas y distribuciones individuales.
# Ideal para analizar interacciones entre múltiples variables.
# Excluimos columnas no numéricas si no son adecuadas para pairplot, o usamos 'hue' si queremos diferenciarlas.
sns.pairplot(df.select_dtypes(include=np.number).dropna()) # pairplot para variables numéricas
plt.suptitle("Pair Plot para Análisis Multivariado", y=1.02) # Ajusta el título para que no se superponga
plt.show()
print("Pair Plot generado para explorar relaciones entre pares de variables numéricas y sus distribuciones.")
print("\n")


# --- 4.4 Visualización de Datos ---
# Las herramientas de visualización ayudan a interpretar los datos de manera más efectiva.
# Incluye histogramas, diagramas de dispersión, gráficos de barras y mapas de calor.
#


# =============================================================================
# 5: Herramientas para EDA
# =============================================================================
"""
--- Módulo 5: Herramientas para EDA ---
Las herramientas más populares para EDA incluyen:
 - Pandas para manipulación de datos tabulares.
 - NumPy para cálculos numéricos.
 - Matplotlib y Seaborn para visualizaciones estadísticas.
 - SciPy y Statsmodels para análisis estadístico avanzado.
Estas bibliotecas de Python se complementan para un ecosistema completo.
"""

# Ejemplo con Statsmodels para análisis estadístico avanzado
# Statsmodels permite aplicar modelos estadísticos como regresiones lineales, ANOVA y pruebas de hipótesis.
# También proporciona herramientas para diagnósticos de regresión, análisis de residuos, intervalos de confianza y tests de significancia.
# Nota: Este es un ejemplo avanzado y puede ser opcional dependiendo del nivel de la audiencia.
# Aquí un ejemplo de regresión lineal simple como se muestra en la fuente.
try:
    import statsmodels.api as sm

    # Eliminar filas con NaN en 'Horas_Estudio' o 'Calificacion' para el modelo
    df_model = df.dropna(subset=['Horas_Estudio', 'Calificacion']).copy()

    X = df_model['Horas_Estudio'] # Variable independiente
    y = df_model['Calificacion'] # Variable dependiente

    X = sm.add_constant(X) # Agrega intercepto
    modelo = sm.OLS(y, X).fit() # Ajusta un modelo de Mínimos Cuadrados Ordinarios (OLS)

    print("\nEjemplo de Análisis Estadístico Avanzado con Statsmodels:")
    print("Resumen estadístico de un modelo de regresión lineal (Horas_Estudio vs Calificacion):")
    print(modelo.summary()) # Ver resumen estadístico del modelo
    print("Statsmodels es útil para aplicar modelos estadísticos y realizar pruebas de hipótesis.")

except ImportError:
    print("\nStatsmodels no está instalado. Omite el ejemplo de análisis estadístico avanzado.")
    print("Puedes instalarlo con: pip install statsmodels")

print("\n")

# =============================================================================
# 6: Aplicaciones del EDA
# =============================================================================

# --- 6: Aplicaciones del EDA ---
# El EDA tiene diversas aplicaciones, incluyendo:
#  - Ciencia de Datos: Preparación de datos para modelos de machine learning, detección de valores atípicos y selección de características.
#  - Negocios y Marketing: Identificación de tendencias en ventas y comportamiento del cliente, segmentación y detección de oportunidades de mercado.
#  - Investigación Científica: Exploración de patrones en datos experimentales, comparación de distribuciones y validación de hipótesis.
#  - Finanzas y Contabilidad: Detección de fraudes, análisis de riesgos financieros, evaluación de portafolios de inversión y seguimiento de tendencias económicas.
#
```
