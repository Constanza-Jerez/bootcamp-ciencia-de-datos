## 1. Concepto de Residuos en Gradient Boosting
Gradient Boosting construye modelos secuencialmente, donde cada nuevo modelo intenta corregir los errores (residuos) de los anteriores.

```python
# Ejercicio 6.1: Concepto de Residuos en Regresión (Base para Gradient Boosting)
y_real_ex = np.array([10, 15, 20, 25, 30])
modelo_inicial_pred = np.array([np.mean(y_real_ex)] * len(y_real_ex))
residuos_f0 = y_real_ex - modelo_inicial_pred
print(f'Valores reales (y_real): {y_real_ex}')
print(f'Predicciones del Modelo Inicial (F0 - la media): {modelo_inicial_pred}')
print(f'Residuos de F0: {residuos_f0}')
```

## 2. Proceso de Entrenamiento y Predicción en Gradient Boosting
Explicación del proceso secuencial, función de pérdida y predicción final.

```python
# Ejercicio 7.1: Simulación de un Paso en el Proceso de Gradient Boosting (Regresión)
y_true = np.array([10, 15, 20, 25, 30])
F0 = np.array([np.mean(y_true)] * len(y_true))
residuos_para_h1 = y_true - F0
h1_predicciones = np.array([-5, 0, 0, 5, 10])
learning_rate = 0.5
F1 = F0 + learning_rate * h1_predicciones
print('Predicción inicial F0 (media):', np.round(F0, 2))
print(f'Residuos para el árbol h1: {residuos_para_h1}')
print(f'Predicciones del Modelo h1: {h1_predicciones}')
print(f'Nueva predicción del conjunto F1: {F1}')
```

## 5. Dificultad de Interpretación y XGBoost
- Modelos difíciles de interpretar.
- XGBoost: implementación optimizada, regularización, manejo de datos faltantes.

### Ejemplo: Uso básico de XGBoost para clasificación

```python
try:
    import xgboost as xgb
    from sklearn.model_selection import train_test_split
    from sklearn.metrics import accuracy_score
    from sklearn.datasets import make_classification
    X_clf_xgb, y_clf_xgb = make_classification(n_samples=1000, n_features=20, n_informative=10, n_redundant=5, random_state=42)
    X_train_xgb, X_test_xgb, y_train_xgb, y_test_xgb = train_test_split(X_clf_xgb, y_clf_xgb, test_size=0.2, random_state=42)
    modelo_xgb_clf = xgb.XGBClassifier(objective='binary:logistic', n_estimators=100, learning_rate=0.1, use_label_encoder=False, eval_metric='logloss', random_state=42)
    modelo_xgb_clf.fit(X_train_xgb, y_train_xgb)
    y_pred_xgb = modelo_xgb_clf.predict(X_test_xgb)
    precision_xgb = accuracy_score(y_test_xgb, y_pred_xgb)
    print(f'Precisión del modelo XGBoost: {precision_xgb:.4f}')
except ImportError:
    print('Instala xgboost y scikit-learn para ejecutar este ejemplo.')
```

Pasos típicos: importar librerías, generar/cargar datos, dividir, entrenar, evaluar, visualizar y mejorar el modelo.

```python
# Paso 1: Importar librerías
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import mean_squared_error, accuracy_score
from sklearn.ensemble import GradientBoostingRegressor, GradientBoostingClassifier
from sklearn.datasets import make_regression, make_classification
# import matplotlib.pyplot as plt

# Paso 2: Generar o cargar datos
X_reg_act, y_reg_act = make_regression(n_samples=500, n_features=5, noise=0.5, random_state=42)
X_clf_act, y_clf_act = make_classification(n_samples=500, n_features=5, n_informative=3, n_redundant=0, random_state=42)
print('Datos sintéticos generados.')

# Paso 3: Dividir los datos
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_reg_act, y_reg_act, test_size=0.2, random_state=42)
X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(X_clf_act, y_clf_act, test_size=0.2, random_state=42)
print('Datos divididos en entrenamiento y prueba.')

# Paso 4: Entrenar modelo de regresión
modelo_gb_reg_act = GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, max_depth=3, random_state=42)
modelo_gb_reg_act.fit(X_train_reg, y_train_reg)
y_pred_reg_act = modelo_gb_reg_act.predict(X_test_reg)
rmse_act = np.sqrt(mean_squared_error(y_test_reg, y_pred_reg_act))
print(f'RMSE en regresión: {rmse_act:.4f}')

# Paso 5: Entrenar modelo de clasificación
modelo_gb_clf_act = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, max_depth=3, random_state=42)
modelo_gb_clf_act.fit(X_train_clf, y_train_clf)
y_pred_clf_act = modelo_gb_clf_act.predict(X_test_clf)
accuracy_act = accuracy_score(y_test_clf, y_pred_clf_act)
print(f'Precisión en clasificación: {accuracy_act:.4f}')

# Paso 6: Mejorar el modelo (GridSearchCV ejemplo conceptual)
param_grid_act = {
    'n_estimators': [50, 100],
    'learning_rate': [0.05, 0.1, 0.2],
    'max_depth': [2, 3, 4]
}
grid_search_reg = GridSearchCV(GradientBoostingRegressor(random_state=42), param_grid_act, cv=3, scoring='neg_mean_squared_error', n_jobs=-1)
grid_search_reg.fit(X_train_reg, y_train_reg)
print(f'Mejores hiperparámetros encontrados: {grid_search_reg.best_params_}')
```
